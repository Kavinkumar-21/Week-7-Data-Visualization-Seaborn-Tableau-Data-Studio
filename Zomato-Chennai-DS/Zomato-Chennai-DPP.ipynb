```python
import pandas as pd
```

```python
dataset = pd.read_csv("Zomato Chennai Listing 2020.csv")
dataset
```

```python
dataset.isna().sum()
```

```python
dataset['Dining Rating'].fillna(0,inplace=True)
dataset['Dining Rating Count'].fillna(0,inplace=True)
dataset['Delivery Rating'].fillna(0,inplace=True)
dataset['Delivery Rating Count'].fillna(0,inplace=True)
```

```python
dataset.isna().sum()
```

```python
dataset
```

```python
def quanqual(dataset):
    quan=[]
    qual=[]
    for columnName in dataset.columns:
        if dataset[columnName].dtype == 'O': 
            qual.append(columnName)
        else:
            quan.append(columnName)
    return quan,qual
```

```python
quan,qual=quanqual(dataset)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='mean')
imp.fit(dataset[quan])
df=imp.transform(dataset[quan])
df
```

```python
df=pd.DataFrame(df,columns = quan)
```

```python
df
```

```python
imp
```

```python
categorical = dataset[qual]
df = dataset[quan]
merge = [categorical,df]
preprocessed = pd.concat(merge,axis=1)

preprocessed
```

```python
preprocessed.to_csv("Preprocessed_Chennai-Zomato-DS.csv",index=False)
```
